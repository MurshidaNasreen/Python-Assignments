```python
#2
strM="Python beginner tutorial"
#Upper
print(strM.upper())
#Lower
print(strM.lower())
#Capitalize
print(strM.capitalize())
#Count t
print(strM.count("t"))
#Replace Python
print(strM.replace("Python","Machine Learning"))
```
