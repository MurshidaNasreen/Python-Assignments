```python
#4.
student_marks = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "Diana": 95,
    "Evan": 64
}
print("student marks")
# b. Access and print the mark of a specific student (e.g., Beena)
specific_student = "Alice"
print(f"The mark of {specific_student} is: {student_marks[specific_student]}")

# c. Add a new student "Janani" with a mark of 80
student_marks["Janani"] = 80

# d. Update the mark of any one older student to 82
student_marks["Bob"] = 82

# Print the final dictionary to verify the changes
print("Updated student_marks dictionary:", student_marks)

# Print all keys using the keys() method
print("--- Keys ---")
print(student_marks.keys())


# 2. Print all values using the values() method
print("\n--- Values ---")
print(student_marks.values())

# 3. Print all key-value pairs using the items() method
print("\n--- Key-Value Pairs ---")
print(student_marks.items())

```

```python
#Sets
my_set = set(["a", 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)

#'set' object does not support item assignment,Sets are Unordered,Sets do not support direct item assignment
# Create the two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Compute union and intersection
set_union = set1.union(set2)
set_intersection = set1.intersection(set2)
# Print the results
print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", set_union)
print("Intersection:", set_intersection)


```
